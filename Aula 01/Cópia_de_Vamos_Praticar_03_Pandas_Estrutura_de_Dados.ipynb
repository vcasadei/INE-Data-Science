- Pandas é uma ferramenta avançada para manipulação de dados, desenvolvida sobre o pacote Numpy. Suas estruturas de dados são poderosas e versáteis, tornando-o amplamente utilizado na área de dados.

## **Estrutura de Dados no Pandas**

### Dataframe

Exemplo Dataframe

[Link Imagem](https://pandas.pydata.org/docs/getting_started/intro_tutorials/01_table_oriented.html)

- O DataFrame é uma das estruturas de dados mais importantes e poderosas da biblioteca Pandas.
- Pode-se pensar em um DataFrame como uma tabela de dados, muito semelhante a uma planilha do Excel ou a uma tabela em um banco de dados.
- É composto por linhas e colunas, onde cada coluna pode conter diferentes tipos de dados (números, strings, datas, etc.).

- Características Principais:
  - Estrutura Bidimensional: Um DataFrame é uma estrutura de dados bidimensional, o que significa que ele tem linhas e colunas. Cada coluna em um DataFrame é uma Series, e as colunas podem ter diferentes tipos de dados.

  - Índice e Colunas: Os DataFrames têm índices para as linhas e rótulos para as colunas. O índice das linhas permite acessar e manipular dados de maneira eficiente, enquanto os rótulos das colunas permitem referenciar cada coluna individualmente.

  - Integração com outras ferramentas: DataFrames podem ser criados a partir de várias fontes de dados, como arquivos CSV, bancos de dados SQL, arquivos Excel, json, entre outros. Além disso, eles podem ser exportados para vários formatos, facilitando a integração com outras ferramentas de análise de dados.

  - Criação de um DataFrame:
    - **df = pd.DataFrame(data)**

  [Saiba mais na documentação oficial](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame)


### Series

Exemplo Series

[Link Imagem](https://pandas.pydata.org/docs/getting_started/intro_tutorials/01_table_oriented.html)

- É uma das estruturas fundamentais da biblioteca Pandas, amplamente utilizada para manipulação e análise de dados.

- Uma Series é essencialmente uma coluna de dados, similar a uma coluna de uma tabela ou uma planilha de Excel. Então, cada coluna em um DataFrame é uma Series.

- Características Principais:
  - Estrutura Unidimensional: Uma Series é um array unidimensional que pode conter qualquer tipo de dado (inteiros, float, strings etc).

  - Indexação: Possui um índice (index) associado a cada elemento. Esse índice é uma sequência de rótulos (labels) que permite o acesso direto aos elementos da Series. Por padrão, esse índice é numérico e começa em 0, mas pode ser personalizado para qualquer tipo de rótulo, como strings ou datas.

- Criação de uma Series:
    - **s = pd.Series(data, index=index)**
      - **data** pode ser um array, número, sequência de números, dicionário etc.
      - se nenhum **index** for passado, será criado um numérico iniciando do 0, de tamanho data - 1.

  [Saiba mais na documentação oficial](https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series)

#### Criando Series

```python
# a partir de um np.array (especificando o index)

import numpy as np # importando a biblioteca numpy

series_a = pd.Series(np.random.random(6), index=["a", "b", "c", "d", "e", "f"]) # especificando o index
series_a # visualizando a Series criada
```

```python
# a partir de um np.array

import numpy as np # importando a biblioteca numpy

series_b = pd.Series(np.random.random(5)) # sem passar o index
series_b # visualizando a Series criada
```

```python
# a partir de um dicionário

dicionario = {"b": 1, "a": 0, "c": 2} # dicionário
# chave = index
# valor =  dados da Series

series_c = pd.Series(dicionario, index = ["b", "a", "c", "f"])
series_c
```

```python
# passando número diferente de valores e indice

pd.Series(dicionario, index=["b", "c", "d", "a"]) # dicionario tem 3 valores e novo index, 4

# NaN (não é um número) é o marcador padrão de dados ausentes usado no Pandas
```

```python
# a partir de uma lista

vendas_semanais = [250, 300, 150, 200, 400, 500, 300]
series_vendas = pd.Series(vendas_semanais)
series_vendas
```

```python
# a partir de uma lista (especificando o index)
vendas_semanais = [250, 300, 150, 200, 400, 500, 300]
dias_semana = ["segunda", "terca", "quarta", "quinta", "sexta", "sabado", "domingo"]
series_vendas = pd.Series(vendas_semanais, index=dias_semana, name="faturamento_por_dia") # name: nome da Series
series_vendas
```

```python
# com condicional
notas = [4.5, 7.0, 5.5, 3.0, 6.0]
status = pd.Series(['Aprovado' if nota >= 5 else 'Reprovado' for nota in notas], name="Resultado")

status
```
