# Leitura de Dados no Pandas

```python
# importação da biblioteca Pandas
import pandas as pd
```

## pd.read_json

```python
json = open('aluguel.json')
print(json.read())
```

```python
df_json = pd.read_json('aluguel.json')
df_json
```

## pd.read_table

```python
txt = open('aluguel.txt')
print(txt.read())
```

```python
df_txt = pd.read_table('aluguel.txt')
df_txt
```

## pd.read_excel

```python
df_xlsx = pd.read_excel('aluguel.xlsx')
df_xlsx
```

## read_html

```python
df_html = pd.read_html('dados_html_1.html')
df_html[0]
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_United_States_Senators_from_Ohio'

# Ler todas as tabelas HTML da página
tabelas = pd.read_html(url)
tabelas[0]
```
