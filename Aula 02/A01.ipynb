```python
df = pd.read_excel('online_retail_II.xlsx')
```

### Inspeção do Dataset

```python
print("Resumo dos Dados Brutos:")
print(df.info())
```

```python
print(df.describe())
```

### Removendo dados Nulos

```python
df_cleaned = df.dropna(subset=['Customer ID'])
print(f"Shape after removing nulls in 'Customer ID': {df_cleaned.shape}")
```

```python
print("Resumo dos Dados após remoção de Nulos:")
print(df_cleaned.info())
print(df_cleaned.describe())
```

### Filtrando Quantidades válidas

Quantidades > 0

```python
df_cleaned = df_cleaned[df_cleaned['Quantity'] > 0]
```

```python
print("Resumo dos Dados após remoção quantidade <= 0:")
print(df_cleaned.info())
print(df_cleaned.describe())
```

```python
df_cleaned['Customer ID'] = df_cleaned['Customer ID'].astype(int)
print("Data type of 'Customer ID' after conversion:")
print(df_cleaned['Customer ID'].dtype)
```

```python
print("Resumo dos Dados após Conversão de tipo do Customer ID")
print(df_cleaned.info())
print(df_cleaned.describe())
```

```python
df_aux = df_cleaned.copy()
df_aux['InvoiceDate'] = df_aux['InvoiceDate'].astype(object)
print("Data type of 'InvoiceDate' after conversion:")
print(df_aux['InvoiceDate'].dtype)
```

```python
df_aux = pd.to_datetime(df_aux['InvoiceDate'])
print("Data type of 'InvoiceDate' after conversion:")
print(df_aux.dtype)
```
